# Where this runs

This is the first page to open when you show CityVigilance, or when you start testing a model.

The command centre and these notebooks use **sample models**. A sample model returns a prepared number. It does not look at a camera picture. It does not use a GPU. It does not call a vision API.

The desk in `command-centre/` is what you show. These notebooks are where you test one model at a time before that model is plugged into the desk.


```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from cityvigilance.runtime import explain, where_this_runs
from cityvigilance.sample import check_event, run_model

print(explain())
```

## The six events

Each model returns one event. The desk only reads that shape. Crowd, movement, and weapon share Crowd camera 1. Traffic and the plate share Junction camera 1.


```python
for name in ("crowd", "movement", "weapon", "accident", "traffic", "plate"):
    event = run_model(name, 0)
    check_event(event)
    print(name, event["cameraId"], event["source"], event["finding"])
print("runtime", where_this_runs()["runs_on"])
```

## What a later model changes

Keep the event shape. Change `source` from `sample` to `local` when the model runs on your machine, or to `api` when a vision service the department has approved does the work. Notebooks `01` to `05` are the benches. Notebook `06` explains demo data, a live file, and why V-JEPA is not the categoriser.
