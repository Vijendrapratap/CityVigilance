## Slot for a real model

Return `crash: false` for an ordinary road scene. Return `crash: true` only for the clip a person should review. A false crash is why the desk asks for a dismiss reason.


```python
def my_accident(sequence: int, at: str) -> dict:
    return run_model("accident", sequence, at)

result = my_accident(1, "2026-11-02T00:00:00Z")
check_event(result)
assert result["finding"]["type"] == "accident"
assert isinstance(result["finding"]["crash"], bool)
print("accident event ok", result["finding"])
```
