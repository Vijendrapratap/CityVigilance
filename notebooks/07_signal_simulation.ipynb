# Signal simulation

Traffic control on the desk is this comparison, not a video. A clip cannot show equal rotation and an AI plan on the same junction.

North is the heavy approach. Rotation gives every arm 15 seconds of green and 3 seconds of clearance. The AI plan serves whichever arm is holding the longest queue, for 8 to 24 seconds. Nothing is written to a controller.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from cityvigilance.signals import build_timeline, summarize
fixed = summarize(build_timeline("fixed"))
ai = summarize(build_timeline("ai"))
print(f"{'':12} {'mean queue':>12} {'peak':>8} {'cleared':>8}")
print(f"{'Rotation':12} {fixed['mean_queue']:12.1f} {fixed['peak_queue']:8.1f} {fixed['departed']:8.1f}")
print(f"{'AI plan':12} {ai['mean_queue']:12.1f} {ai['peak_queue']:8.1f} {ai['departed']:8.1f}")
print("north queue at 3:00", round(build_timeline('fixed')[-1]['queues']['N'], 1), "rotation,", round(build_timeline('ai')[-1]['queues']['N'], 1), "AI")
```

The same function drives the Traffic control screen. Over three minutes the AI plan keeps a lower mean queue, a lower peak, and clears more vehicles, because the north approach is not left waiting for its turn.
